```python
import numpy as np

# Parameters
mean = 0.4
variance = 2
std_dev = np.sqrt(variance)

# Number of simulations
n = 1000

# Monte Carlo Simulation
np.random.seed(42)          # Makes results reproducible
tumour = np.random.normal(mean, std_dev, n)

# Probability that tumour value is below 2
probability = np.mean(tumour < 2)

print("Estimated Probability:", probability)
```
